```python
import re

epochs = []
train_loss = []
train_kl = []
train_topk = []
train_cross_entropy = []
val_loss = []
val_kl = []
val_topk = []
val_cross_entropy = []

file_path = r"C:\Users\Tristan\Downloads\HyPCAR3\visuals\finalBaseAbundanceTraining.txt"

with open(file_path, "r") as f:
    lines = f.readlines()

for i in range(0, len(lines), 2):
    #Remove any extra whitespace or newline characters
    train_line = lines[i].strip()
    val_line = lines[i + 1].strip()

    #Use regular expressions to parse the training metrics
    #Expected format:
    #"Epoch 1, Loss: 0.020120840519666672, KL Divergence: 0.3243209423886405, Top-K Accuracy: 0.4519097222222222, Cross Entropy: 1.3845441341400146"
    train_pattern = (r"Epoch (\d+), Loss: ([\d\.eE\-\+]+), "
                     r"KL Divergence: ([\d\.eE\-\+]+), Top-K Accuracy: ([\d\.eE\-\+]+), "
                     r"Cross Entropy: ([\d\.eE\-\+]+)")
    train_match = re.match(train_pattern, train_line)
    if not train_match:
        print("Could not parse training line:", train_line)
        continue#Skip to the next pair

    epoch = int(train_match.group(1))
    loss = float(train_match.group(2))
    kl = float(train_match.group(3))
    topk = float(train_match.group(4))
    cross = float(train_match.group(5))
    
    epochs.append(epoch)
    train_loss.append(loss)
    train_kl.append(kl)
    train_topk.append(topk)
    train_cross_entropy.append(cross)

    #"Validation Loss: 0.020075684413313866, KL Divergence: 0.31906323426040833, Top-K Accuracy: 0.4634214031971581, Cross Entropy: 1.375912070274353"
    val_pattern = (r"Validation Loss: ([\d\.eE\-\+]+), "
                   r"KL Divergence: ([\d\.eE\-\+]+), Top-K Accuracy: ([\d\.eE\-\+]+), "
                   r"Cross Entropy: ([\d\.eE\-\+]+)")
    val_match = re.match(val_pattern, val_line)
    if not val_match:
        print("Could not parse validation line:", val_line)
        continue

    val_loss_val = float(val_match.group(1))
    val_kl_val = float(val_match.group(2))
    val_topk_val = float(val_match.group(3))
    val_cross = float(val_match.group(4))

    val_loss.append(val_loss_val)
    val_kl.append(val_kl_val)
    val_topk.append(val_topk_val)
    val_cross_entropy.append(val_cross)


import matplotlib.pyplot as plt
plt.figure(figsize=(12, 3.5))
plt.plot(epochs, train_loss, label="Train MSE Loss")
plt.plot(epochs, val_loss, label="Validation MSE Loss", linestyle="--")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("MSE Loss Over Epochs")
plt.legend()
plt.show()


plt.figure(figsize=(12, 3.5))
plt.plot(epochs, train_kl, label="Train KL-Divergence")
plt.plot(epochs, val_kl, label="Validation KL-Divergence", linestyle="--")
plt.xlabel("Epoch")
plt.ylabel("KL-Divergence")
plt.title("KL-Divergence Over Epochs")
plt.legend()
plt.show()

plt.figure(figsize=(12, 3.5))
plt.plot(epochs, train_cross_entropy, label="Train Cross-Entropy")
plt.plot(epochs, val_cross_entropy, label="Validation Cross-Entropy", linestyle="--")
plt.xlabel("Epoch")
plt.ylabel("Cross-Entropy")
plt.title("Cross-Entropy Over Epochs")
plt.legend()
plt.show()

plt.figure(figsize=(12, 3.5))
plt.plot(epochs, train_topk, label="Train Top-1 Accuracy")
plt.plot(epochs, val_topk, label="Validation Top-1 Accuracy", linestyle="--")
plt.xlabel("Epoch")
plt.ylabel("Top-1 Accuracy")
plt.title("Top-1 Accuracy Over Epochs")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

C_fixed=0.2
N_fixed=0.2

#Create a grid over H and O
tau = 1#For visual purposes
H_vals=np.linspace(0.0, 1.0, 300)#total H budget
O_vals=np.linspace(0.0, 1.0, 300)#total O budget
H, O=np.meshgrid(H_vals, O_vals)

#Figure out different atmospheric type thresholds
condA=H-(2*O + 4*C_fixed)
condA_gate=1/(1 + np.exp(-condA / tau))

condA2=condA - (3*N_fixed - (H - 2*O - 4*C_fixed))

condB=2*O - (H + 4*C_fixed)

condC_val=np.abs(H + C_fixed + O + N_fixed - 1.0)

#Turn each into a soft gate
σ=lambda x: 1/(1 + np.exp(-x/tau))

gateA1=condA_gate * σ(condA2)
gateA2=condA_gate * (1 - σ(condA2))
gateB =(1 - condA_gate) * σ(condB)
gateC =(1 - condA_gate) * (1 - gateB) * σ(-condC_val)
gateU =np.clip(1 - (gateA1 + gateA2 + gateB + gateC), 0, 1)

gates = {
    'Gate A1': gateA1,
    'Gate A2': gateA2,
    'Gate B' : gateB,
    'Gate C' : gateC,
    'Gate U' : gateU
}


fig, axes=plt.subplots(2, 3, figsize=(12, 8))
fig.subplots_adjust(wspace=0.75, hspace=0.5)#Adjust horizontal & vertical spacing
axes=axes.flatten()

for ax,(name, gate) in zip(axes, gates.items()):
    im = ax.pcolormesh(H_vals, O_vals, gate, shading='auto')
    ax.set_title(name,fontsize=18)
    ax.set_xlabel('H budget',fontsize=15)
    ax.set_ylabel('O budget',fontsize=15)
    ax.tick_params(axis="both",labelsize=15)
    fig.colorbar(im, ax=ax).set_label(f'{name} value', fontsize=18) 

axes[-1].axis('off')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
from itertools import product
from itertools import product

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

from sklearn import metrics, neighbors
from sklearn.preprocessing import MinMaxScaler


def predicted_vs_real(y_real, y_pred, names, ranges):

    num_plots = y_pred.shape[1]
    num_plot_rows = int(np.sqrt(num_plots))
    num_plot_cols = (num_plots - 1) // num_plot_rows + 1
    
    fig, axes = plt.subplots(num_plot_rows, num_plot_cols,
                             figsize=(5*num_plot_cols, 5*num_plot_rows))
    
    for dim, (ax, name_i, range_i) in enumerate(zip(axes.ravel(), names, ranges)):
        
        current_real = y_real[:, dim]
        current_pred = y_pred[:, dim]
        
        r2 = metrics.r2_score(current_real, current_pred)
        label = "$R^2 = {:.3f}$".format(r2)
        ax.plot(current_real, current_pred, '.', label=label)
        
        ax.plot(range_i, range_i, '--', linewidth=3, color="C3")
        
        ax.axis("equal")
        ax.grid()
        ax.set_xlim(range_i)
        ax.set_ylim(range_i)
        ax.set_xlabel("Real {}".format(name_i), fontsize=18)
        ax.set_ylabel("Predicted {}".format(name_i), fontsize=18)
        ax.legend(loc="upper left", fontsize=14)
    
    fig.tight_layout()
    return fig


def feature_importances(forests, names, colors):
    
    num_plots = len(forests)
    num_plot_rows = (num_plots - 1) // 2 + 1
    num_plot_cols = 2
    
    fig, axes = plt.subplots(num_plot_rows, num_plot_cols,
                             figsize=(15, 3.5*num_plot_rows))
    
    for ax, forest_i, name_i, color_i in zip(axes.ravel(), forests, names, colors):
        ax.bar(np.arange(len(forest_i.feature_importances_)), forest_i.feature_importances_,
               label="Importance for {}".format(name_i),
               width=0.4, color=color_i)
        ax.set_xlabel("Feature index", fontsize=18)
        ax.legend(fontsize=16)
        ax.grid()
    
    fig.tight_layout()
    return fig

def posterior_matrix(estimations, y, names, ranges, colors, soft_colors=None):
    cmaps = [LinearSegmentedColormap.from_list("MyReds", [(1, 1, 1), c], N=256) for c in colors]
    ranges = np.array(ranges)

    if soft_colors is None:
        soft_colors = colors

    num_dims = estimations.shape[1]
    fig, axes = plt.subplots(nrows=num_dims, ncols=num_dims,
                             figsize=(2 * num_dims, 2 * num_dims))
    # fig.subplots_adjust(left=0.07, right=1-0.05,
    #                     bottom=0.07, top=1-0.05,
    #                     hspace=0.05, wspace=0.05)
    fig.subplots_adjust(left=0.07, right=0.95, bottom=0.07, top=0.95,
                        hspace=0.05, wspace=0.05)
    fig.suptitle("HyPCAR Posterior Predictions",fontsize=22,y=1)    
    for ax, dims in zip(axes.flat, product(range(num_dims), range(num_dims))):
        dims = list(dims[::-1])
        ax.xaxis.set_visible(False)
        ax.yaxis.set_visible(False)
        ax.title.set_visible(False)

        # Enable axes ticks and labels for boundary subplots (first row/column, etc.)
        if dims[1] == 0:
            ax.xaxis.set_visible(True)
            ax.xaxis.set_ticks_position('top')
            if names is not None:
                ax.set_xlabel(names[dims[0]], fontsize=18, labelpad=12)
        # For the bottom row (row index == num_dims - 1), show the x-axis at the bottom.
        if dims[1] == num_dims - 1:
            ax.xaxis.set_visible(True)
            ax.xaxis.set_ticks_position('bottom')
            if names is not None:
                ax.set_xlabel(names[dims[0]], fontsize=18, labelpad=12)
        # For the left column (column index == 0), show the y-axis on the left and add labels.
        if dims[0] == 0:
            ax.yaxis.set_visible(True)
            ax.yaxis.set_ticks_position('left')
            if names is not None:
                ax.set_ylabel(names[dims[1]], fontsize=18, labelpad=12)
        # For the right column (column index == num_dims - 1), show the y-axis on the right.
        if dims[0] == num_dims - 1:
            ax.yaxis.set_visible(True)
            ax.yaxis.set_ticks_position('right')



        if dims[0] < dims[1]:  # Off-diagonal upper triangle
            locations, kd_probs, *_ = _kernel_density_joint(estimations[:, dims], ranges[dims])
            ax.contour(locations[0], locations[1], kd_probs, colors=colors[dims[0]], linewidths=0.5)
            histogram, grid_x, grid_y = _histogram(estimations[:, dims], ranges[dims])
            ax.pcolormesh(grid_x, grid_y, histogram, cmap=cmaps[dims[0]])

            expected = np.median(estimations[:, dims], axis=0)
            ax.plot([expected[0], expected[0]], [ranges[dims[1]][0], ranges[dims[1]][1]], '-', linewidth=1, color='#222222')
            ax.plot([ranges[dims[0]][0], ranges[dims[0]][1]], [expected[1], expected[1]], '-', linewidth=1, color='#222222')
            ax.plot(expected[0], expected[1], '.', color='#222222')
            if y is not None:
                real = y[dims]
                ax.plot(real[0], real[1], '*', markersize=10, color='#FF0000')
            ax.axis([ranges[dims[0]][0], ranges[dims[0]][1], ranges[dims[1]][0], ranges[dims[1]][1]])
        elif dims[0] > dims[1]:  # Off-diagonal lower triangle scatter plot
            ax.plot(estimations[:, dims[0]], estimations[:, dims[1]], '.', color=soft_colors[dims[1]])
            ax.axis([ranges[dims[0]][0], ranges[dims[0]][1], ranges[dims[1]][0], ranges[dims[1]][1]])
        else:  # Diagonal (histogram)
            histogram, bins = np.histogram(estimations[:, dims[0]], bins=20, range=ranges[dims[0]])
            ax.bar(bins[:-1], histogram, color=soft_colors[dims[0]], width=bins[1]-bins[0])

            expected = np.median(estimations[:, dims[0]])
            ax.plot([expected, expected], [0, 1.1 * max(histogram)], '-', linewidth=1, color='#222222')
            if y is not None:
                real = y[dims[0]]
                ax.plot([real, real], [0, max(histogram)], 'r-')
            ax.axis([ranges[dims[0]][0], ranges[dims[0]][1], 0, 1.1 * max(histogram)])
    return fig


def _min_max_scaler(ranges, feature_range=(0, 100)):
    res = MinMaxScaler()
    res.data_max_ = ranges[:, 1]
    res.data_min_ = ranges[:, 0]
    res.data_range_ = res.data_max_ - res.data_min_
    res.scale_ = (feature_range[1] - feature_range[0]) / (ranges[:, 1] - ranges[:, 0])
    res.min_ = -res.scale_ * res.data_min_
    res.n_samples_seen_ = 1
    res.feature_range = feature_range
    return res


def _kernel_density_joint(estimations, ranges, bandwidth=1/25):
    
    ndims = len(ranges)
    
    scaler = _min_max_scaler(ranges, feature_range=(0, 100))
    
    bandwidth = bandwidth * 100
    # step = 1.0
    
    kd = neighbors.KernelDensity(bandwidth=bandwidth).fit(scaler.transform(estimations))
    locations1d = np.arange(0, 100, 1)
    locations = np.reshape(np.meshgrid(*[locations1d] * ndims), (ndims, -1)).T
    kd_probs = np.exp(kd.score_samples(locations))
    
    shape = (ndims,) + (len(locations1d),) * ndims
    locations = scaler.inverse_transform(locations)
    locations = np.reshape(locations.T, shape)
    kd_probs = np.reshape(kd_probs, shape[1:])
    return locations, kd_probs, kd


def _histogram(estimations, ranges, bins=20):
    
    if len(ranges) == 1:
        histogram, edges = np.histogram(estimations[:, 0], bins=bins, range=ranges[0])
        return histogram, edges
    
    if len(ranges) == 2:
        histogram, xedges, yedges = np.histogram2d(estimations[:, 0], estimations[:, 1], bins=bins, range=ranges)
        grid_x, grid_y = np.meshgrid(xedges, yedges)
        return histogram.T, grid_x, grid_y,  

means = np.array([0.1154, 0.5474, 0.0, 0.1856, 0.1517, 0.0, 0.0])
sigmas = np.array([0.0730, 0.2337, 0.1851, 0.0588, 0.0490, 0.0733, 0.1260])
true_vals = np.array([0.0449, 0.5785, 0.0, 0.2153, 0.1613, 0.0, 0.0])

#Generate posterior samples
n_samples = 1000  
estimations = np.random.normal(loc=means, scale=sigmas, size=(n_samples, len(means)))

#Define ranges based on means and sigmas
ranges = [
    (mean - 3 * sigma, mean + 3 * sigma) 
    for mean, sigma in zip(means, sigmas)
]

colors = ['#1f77b4','#ff7f0e','#2ca02c','#d62728','#9467bd','#8c564b','#e377c2']

names = ["O2","N2","H2","CO2","H2O","CH4","NH3"]



fig = posterior_matrix(
    estimations=estimations,       
    y=true_vals,                   
    names=names,                   
    ranges=ranges,                 
    colors=colors                  
)


plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import truncnorm
import matplotlib.gridspec as gridspec

means  = np.array([0.0344, 0.1852, 0.0820, 0.0161, 0.2792, 0.1918, 0.2113])
sigmas = np.array([0.2582, 0.2176, 0.5122, 0.1579, 0.2827, 0.1204, 0.3484])
names  = ["O2", "N2", "H2", "CO2", "H2O", "CH4", "NH3"]
colors = ['#1f77b4','#ff7f0e','#2ca02c','#d62728','#9467bd','#8c564b','#e377c2']

def sample_truncnorm(mu, sigma, N, low=0.0, high=1.0):
    a, b = (low - mu) / sigma, (high - mu) / sigma
    return truncnorm(a, b, loc=mu, scale=sigma).rvs(size=N)

# Monte Carlo sampling
N = 2000
samples = np.vstack([sample_truncnorm(means[i], sigmas[i], N)
                     for i in range(len(means))]).T

ci95 = np.array([np.percentile(samples[:, i], [2.5, 97.5]) for i in range(len(means))])

fig = plt.figure(figsize=(14, 6))
gs = gridspec.GridSpec(2, 4, height_ratios=[1, 1], wspace=0.3, hspace=0.4)

axes = []
for col in range(4):
    axes.append(fig.add_subplot(gs[0, col]))
fig.add_subplot(gs[0, 3]).axis('off') 

for col in range(3):
    axes.append(fig.add_subplot(gs[1, col]))

for i, ax in enumerate(axes):
    if i >= len(means):
        break
    # shade CI
    lo, hi = ci95[i]
    ax.axvspan(lo, hi, color=colors[i], alpha=0.2)

    counts, bins = np.histogram(samples[:, i], bins=20, density=True)
    centers = 0.5 * (bins[:-1] + bins[1:])
    ax.fill_between(centers, counts, color=colors[i], alpha=0.6)
    ax.plot(centers, counts, color='k', lw=1)

    ax.axvline(means[i], color='red', linestyle='--', lw=2)
    height = np.interp(means[i], centers, counts)
    ax.plot(means[i], height, 'r*', ms=10, clip_on=False)

    ax.set_xlim(ci95[i])
    ax.set_xlabel(names[i], fontsize=12)

    ax.grid(axis='x', linestyle='--', alpha=0.3)

axes[0].set_ylabel("Density", fontsize=12)

fig.suptitle("LHS 1140 b Posterior Marginals\n95% CI Shading & Predicted Means", fontsize=16, y=1.02)

plt.tight_layout(rect=[0,0,1,0.95])
plt.show()
```
